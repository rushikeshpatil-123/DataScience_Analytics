```python
import numpy as np
import pandas as pd
```

```python
# 1D
array =pd.Series(np.arange(1,10,1))
# 2D : Multi Dimensional
array1 = pd.DataFrame({
    "Name":[np.nan, "Tathagat","Rushikesh","Shivam","Sanjeev",np.nan],
    "CGPA":[10,8,9,np.nan,np.nan,9],
    "Salary":[1000,2000,3000,4000,5000,6000],
    "Age":[19,21,23,25,27,29]
})
```

```python
# Checking Descriptive stats
array1.describe()
```

```python
numerical_col = array1.select_dtypes(exclude = 'object')
categorical_col = array1.select_dtypes(include = 'object')

from collections import OrderedDict
def descriptive_stats():    # Function Defination
  stats = []
  for i in numerical_col.columns:
      numerical_stats = OrderedDict({
          "Feature": i,
          "Count":array1[i].count(),
          "Mean":array1[i].mean(),
          "Std":array1[i].std(),
          "Min":array1[i].min(),
          "Max":array1[i].max(),
          "25%":array1[i].quantile(0.25),
          "50%":array1[i].quantile(0.50),
          "75%":array1[i].quantile(0.75),
          "IQR":array1[i].quantile(0.75)-array1[i].quantile(0.25),
          "Skewness":array1[i].skew(),
          "Kurtosis":array1[i].kurtosis()

      })

      stats.append(numerical_stats)
      report = pd.DataFrame(stats)
  return report

data_info = descriptive_stats()    # Function Calling
print(data_info)
```
